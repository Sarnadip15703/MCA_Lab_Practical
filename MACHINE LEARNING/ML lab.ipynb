# Normalize your data

```python
x1=(x-np.mean(x))/np.std(x)
y1=(y-np.mean(y))/np.std(y)
```

```python
def grad_beta0(beta0,beta1,x1,y1):
    n=len(x1)
    res=0
    for i in range(n):
        res+=(y1[i]-beta0-beta1*x1[i])
    grad_b0=-2/float(n)*res
    #print(grad_b0)
    return grad_b0
```
