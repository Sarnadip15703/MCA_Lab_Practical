```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print(cv2.__version__)
```

```python
img = cv2.imread("img.png", cv2.IMREAD_GRAYSCALE)

# Check whether image was successfully loaded
if img is None:
    print("Error: Image not found!")
    exit()

# Image height and width
height, width = img.shape

# Number of pixels
num_pixels = img.size

# Datatype
datatype = img.dtype

# Minimum pixel intensity
min_intensity = np.min(img)

# Maximum pixel intensity
max_intensity = np.max(img)

# Mean intensity
mean_intensity = np.mean(img)

# Pixel value at specified coordinate
# Example: x = 100, y = 50
x = 100
y = 50

if y < height and x < width:
    pixel_value = img[y, x]
else:
    pixel_value = "Coordinate outside image"

# Display information
print("Image Height:", height)
print("Image Width:", width)
print("Number of Pixels:", num_pixels)
print("Datatype:", datatype)
print("Minimum Pixel Intensity:", min_intensity)
print("Maximum Pixel Intensity:", max_intensity)
print("Mean Intensity:", mean_intensity)
print("Pixel Value at (x=100, y=50):", pixel_value)


# Display using OpenCV
cv2.imshow("Grayscale Image - OpenCV", img)
cv2.waitKey(0)
cv2.destroyAllWindows()


# Display using Matplotlib
plt.imshow(img, cmap="gray")
plt.title("Grayscale Image - Matplotlib")
plt.axis("off")
plt.show()
```
