```python
import cv2
import numpy as np5
import matplotlib.pyplot as plt
```

```python

img1 = cv2.imread("image1.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("image1flip.jpg", cv2.IMREAD_GRAYSCALE)

# Check if images loaded
if img1 is None:
    print("Image 1 not found")
elif img2 is None:
    print("Image 2 not found")
else:

    # Make image 2 the same size as image 1
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

    # Convert both images to binary
    _, img1 = cv2.threshold(img1, 127, 255, cv2.THRESH_BINARY)
    _, img2 = cv2.threshold(img2, 127, 255, cv2.THRESH_BINARY)
    # img1_rgb = cv2.cvtColor(img1, cv2.COLOR_GRAY2RGB)
    # img2_rgb = cv2.cvtColor(img2, cv2.COLOR_GRAY2RGB)

    # AND
    and_img = cv2.bitwise_and(img1, img2)

    # OR
    or_img = cv2.bitwise_or(img1, img2)

    # XOR
    xor_img = cv2.bitwise_xor(img1, img2)

    # NOR = NOT(OR)
    nor_img = cv2.bitwise_not(or_img)

    # Display
    plt.figure(figsize=(15, 8))

    plt.subplot(2, 3, 1)
    plt.imshow(img1, cmap="gray")
    plt.title("Binary Image 1")
    plt.axis("off")

    plt.subplot(2, 3, 2)
    plt.imshow(img2, cmap="gray")
    plt.title("Binary Image 2")
    plt.axis("off")

    plt.subplot(2, 3, 3)
    plt.imshow(and_img, cmap="gray")
    plt.title("AND")
    plt.axis("off")

    plt.subplot(2, 3, 4)
    plt.imshow(or_img, cmap="gray")
    plt.title("OR")
    plt.axis("off")

    plt.subplot(2, 3, 5)
    plt.imshow(xor_img, cmap="gray")
    plt.title("XOR")
    plt.axis("off")

    plt.subplot(2, 3, 6)
    plt.imshow(nor_img, cmap="gray")
    plt.title("NOR")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
