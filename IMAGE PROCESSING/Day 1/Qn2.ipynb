```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image in grayscale
img = cv2.imread("photo.jpg", cv2.IMREAD_GRAYSCALE)

# Check if image is loaded
if img is None:
    print("Image not found!")
    exit()

# --------------------------------
# 1. Increase Brightness
# --------------------------------
bright = np.clip(
    img.astype(np.int16) + 50,
    0,
    255
).astype(np.uint8)


# --------------------------------
# 2. Decrease Brightness
# --------------------------------
dark = np.clip(
    img.astype(np.int16) - 50,
    0,
    255
).astype(np.uint8)


# --------------------------------
# 3. Increase Contrast
# --------------------------------
contrast_high = np.clip(
    img.astype(np.float32) * 1.5,
    0,
    255
).astype(np.uint8)


# --------------------------------
# 4. Decrease Contrast
# --------------------------------
contrast_low = np.clip(
    img.astype(np.float32) * 0.5,
    0,
    255
).astype(np.uint8)


# --------------------------------
# 5. Generate Negative Image
# --------------------------------
negative = 255 - img


# --------------------------------
# 6. Apply Binary Thresholding
# --------------------------------
threshold = 127

binary = np.where(
    img >= threshold,
    255,
    0
).astype(np.uint8)


# --------------------------------
# Display Original and Results
# --------------------------------

plt.figure(figsize=(12, 8))

plt.subplot(2, 4, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(bright, cmap="gray")
plt.title("Increase Brightness")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(dark, cmap="gray")
plt.title("Decrease Brightness")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(contrast_high, cmap="gray")
plt.title("Increase Contrast")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(contrast_low, cmap="gray")
plt.title("Decrease Contrast")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(negative, cmap="gray")
plt.title("Negative Image")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(binary, cmap="gray")
plt.title("Binary Threshold")
plt.axis("off")

plt.tight_layout()
plt.show()
```
